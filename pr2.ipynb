```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
!pip install --upgrade seaborn
# Load dataset
df = sns.load_dataset("iris")

# 1. Display first 5 rows
print("--- First 5 Rows ---")
print(df.head())

# 2. Data info
print("\n--- Data Info ---")
print(df.info())

# 3. Summary statistics
print("\n--- Summary Statistics ---")
print(df.describe())

# 4. Histogram
plt.hist(df['sepal_length'])
plt.title("Sepal Length Distribution")
plt.show()

# 5. Scatter plot
plt.scatter(x='sepal_length', y='sepal_width', data=df)
plt.title("Sepal Length vs Width")
plt.show()

# 6. Heatmap
sns.heatmap(df.corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Create sample tabular dataset
data = {
    'Name': ['Hill1', 'Hill2', 'Hill3', 'Hill4', 'Hill5'],
    'Height': [900, 1100, 850, 1200, 950],
    'Region': ['North', 'South', 'North', 'South', 'East']
}

df = pd.DataFrame(data)

# Step 2: Display dataset
print("--- Dataset Snapshot ---")
print(df)

# Step 3: Summary statistics
print("\n--- Summary Statistics ---")
print(df.describe())

# Step 4: Grouping (Average height per region)
print("\n--- Region-wise Average Height ---")
print(df.groupby('Region')['Height'].mean())

# Step 5: Boxplot visualization
df.boxplot(column='Height')
plt.title("Distribution of Heights")
plt.ylabel("Meters")
plt.show()

# Pie Chart: Count of hills per region
region_counts = df['Region'].value_counts()

plt.figure()
plt.pie(region_counts, labels=region_counts.index, autopct='%1.1f%%', startangle=90)
plt.title("Distribution of Hills by Region")
plt.show()

# Bar Chart: Heights of each hill
plt.figure()
plt.bar(df['Name'], df['Height'], color='skyblue')
plt.title("Height of Hills")
plt.xlabel("Hill Name")
plt.ylabel("Height (Meters)")
plt.show()
```
